## Git in one slide

Git takes **snapshots** of a folder over time.

- Each snapshot is a **commit**, with a message saying what changed
- Nothing is lost; you can return to any snapshot
- Two people can change the same project and git merges it

**GitHub** stores a copy others can reach (a remote). Git works without it; we use both.

## Why a fork and not just a clone

A **clone** is a copy on your laptop. A **fork** is a copy of the whole repo under your own GitHub account.

- The fork is where your work lives, yours and visible to me
- It keeps a link back to mine, so pulling my updates stays easy all term

## The three copies of everything

My repository (upstream) forks to your fork (origin), which clones to your laptop; you pull from upstream and push to origin

- Pull from me; never push to me
- Push to your fork; that fork is what you submit

## The everyday git cycle

Working directory to staging with git add, to local repo with git commit, to your fork on GitHub with git push; git pull brings commits back down

- Edit, snapshot, share. The same few commands, over and over.
- A **commit** saves on your laptop; a **push** puts it on GitHub. Until you push, only you can see it.
- It is your own copy: commit your notes, problem sets, experiments, not only what you hand in.

## Stage and commit

Check what changed, mark what goes into the snapshot, then take it:

```bash
git status                      # what has changed, and where you stand
git add 02-my-notes.ipynb       # stage one file for the next snapshot
git add .                       # or stage everything that changed
git commit -m "Add notes from the regression lecture"
```

- **Staging** is you choosing what goes in each commit. Change ten files, commit only the three that belong together.
- `git status` is always safe and tells you where you stand. Run it constantly.

## A good commit message

The message is a note to your future self, and to me when I read your work. Make it worth reading.

- One short summary line, present tense, saying what the commit does: "Add", "Fix", "Rewrite"
- Describe the change and why, not a line-by-line account
- One logical change per commit, so the history stays readable

| instead of | write |
|---|---|
| `stuff` | `Fix wage column name in problem set 1` |
| `update` | `Add cross validation to question 3` |
| `asdf` | `Rewrite the gradient descent loop in vector form` |

## Pull and push

Committing saves your work on your laptop. Sharing it takes one more step.

```bash
git push origin main    # send your commits up to your fork
git pull origin main    # bring commits down from your fork
```

- **push** sends your snapshots up to GitHub. **pull** brings snapshots down. `main` is the default branch, the main line of your work.
- A good rhythm: pull when you sit down, commit as you go, push when you stand up.
- Push often. A commit that lives only on your laptop is one spilled coffee from gone.

## Branches

A **branch** is a separate line of work. You switch between branches freely, and changes on one do not touch another until you merge.

- `main` is the branch you trust, the version that always works
- Branch off `main` to work without risking it:
    - `experiment`: try an idea you might throw away
    - `ps1`: one assignment, kept to itself
    - `fix-plot`: a single change you can review on its own
- `git checkout -b NAME` makes one and switches to it; `git checkout main` goes back

Nothing you do on a branch touches `main` until you choose to merge it.

## Getting a branch back into main

A ps1 branch splits off main, gets two commits, and merges back into main at a merge commit

- **Fast-forward:** `main` has not moved, so it just slides up to your branch. Straight-line history.
- **Merge commit:** both moved, so git joins them with a commit that has two parents.
- **Pull request:** the same merge, proposed and reviewed on GitHub before it lands. How you merge into a repo you do not own, like mine.

Locally: `git checkout main` then `git merge ps1`. Across repos: open a pull request instead.

## Pull requests

A **pull request** proposes merging one branch into another and asks the owner to accept it. A merge you request rather than do yourself.

- Used when you should not push straight to the target: someone else's repo, or a shared `main` a team protects
- The reviewer sees every change, comments on the exact lines, and merges or closes it
- Nothing lands until they accept, so you cannot break the target

In this course you open a pull request from your fork's branch into my `main`. That is how you hand work in (details later).

## .gitignore

A **`.gitignore`** file lists paths git should not track: scratch data, build output, secrets, notebook checkpoints.

- One pattern per line: `*.csv`, `__pycache__/`, `.ipynb_checkpoints/`
- Ignored files never appear in `git status` and are skipped by `git add .`
- Keeps commits to what matters, and keeps big or private files off GitHub

Expected a file to commit and it did not show up? Something may be ignoring it. `git check-ignore -v THEFILE` names the rule.

## A first plot

If a picture appears below this cell, your plotting is set up correctly.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["education"], df["wage"])

ax.set_xlabel("years of education")
ax.set_ylabel("")
ax.text(0, 1.03, "wage", transform=ax.transAxes, ha="left", va="bottom")
ax.spines[["top", "right"]].set_visible(False)

plt.show()
````

````python
import matplotlib.pyplot as plt
from pywrangling.graphing_functions import use_plotplainblind, set_top_ylabel

use_plotplainblind(kind="scatter")   # plain, colorblind-safe scheme; call once before plotting

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["education"], df["wage"])

ax.set_xlabel("years of education")
set_top_ylabel(ax, "wage")           # horizontal label above the axis, the way I do it

plt.show()
````

The y label sits flat above the axis, not rotated. Rotated text is harder to read, and I do this on every figure. Whether that upward slope is the *return* to education is the whole semester's question.
