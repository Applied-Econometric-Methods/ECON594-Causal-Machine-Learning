### Question 1: Why Bother

Least squares gives the coefficients of a linear regression from a formula.
Why does anything need gradient descent?

- <b style="color:#73000A;">A.</b> Least squares is slow, and gradient descent gets the same answer faster
- <b style="color:#73000A;">B.</b> Plenty of models, logistic regression among them, have no closed-form solution
- <b style="color:#73000A;">C.</b> Least squares only works when the sample is large
- <b style="color:#73000A;">D.</b> Gradient descent is more accurate than least squares

### Question 3: One Sample Per Step

Stochastic gradient descent randomly picks <b style="color:#73000A;">one sample for each step</b> and uses just
that one to calculate the derivatives.

When is that worth doing?

- <b style="color:#73000A;">A.</b> When the loss has no derivative
- <b style="color:#73000A;">B.</b> When there is tons of data and lots of parameters, so regular
  gradient descent is not computationally feasible
- <b style="color:#73000A;">C.</b> When the learning rate is too large to converge
- <b style="color:#73000A;">D.</b> When you want a more accurate answer than regular gradient descent
