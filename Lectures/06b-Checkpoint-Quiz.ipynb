- <b style="color:#73000A;">A.</b> All four with squares and interactions, because 2.60 is the smallest number on the table
- <b style="color:#73000A;">B.</b> Education alone, because the simplest model is the safest
- <b style="color:#73000A;">C.</b> Education, experience, tenure and union, because 3.88 is the smallest error on workers the model never saw

### Question 2: Randomized Experiment

You run a randomized experiment and estimate the treatment effect. A colleague asks for its cross validation error.

What do you tell them?

- <b style="color:#73000A;">A.</b> Run five-fold and report the average
- <b style="color:#73000A;">B.</b> Cross validation answers a prediction question. The treatment effect is an estimate, and it gets a standard error
- <b style="color:#73000A;">C.</b> Cross validation only works on continuous outcomes, so it depends on the outcome
- <b style="color:#73000A;">D.</b> Run it, but only on the treated group
