### Question 1: Which Penalty

Fifty observations, forty candidate regressors, and you have good reason to think
<b style="color:#73000A;">all of them matter a little</b>.

Ridge or lasso? (Elastic net is not available)

- <b style="color:#73000A;">A.</b> Lasso
- <b style="color:#73000A;">B.</b> Ridge
- <b style="color:#73000A;">C.</b> Either one, they give the same answer with the same $\lambda$
- <b style="color:#73000A;">D.</b> Neither. With 40 regressors and 50 observations, least squares is fine

### Question 2: One Month, Two Coefficients

One month of data: 2 on ads, 1 on displays, sales up 5. So any pair with
$2b_{\text{ads}} + b_{\text{display}} = 5$ reproduces the month, and
there are infinitely many of them.

Add the ridge penalty and you get only one answer. Why?

- <b style="color:#73000A;">A.</b> Along the line of perfect fits the penalty works out to $5 + 5t^2$, and a
  parabola has one bottom
- <b style="color:#73000A;">B.</b> Because ridge drops one of the two regressors, which leaves one equation
  and one unknown
- <b style="color:#73000A;">C.</b> Because the software searches the pairs in order and stops at the first
  one that fits
- <b style="color:#73000A;">D.</b> Because only one of those pairs has both coefficients positive

### Question 3: Turning the Dial

You fit ridge to the same data at $\lambda = 0$, then $\lambda = 1$, then
$\lambda = 1000$, and you never run cross validation.

Which statement is right?

- <b style="color:#73000A;">A.</b> At $\lambda = 0$ ridge is undefined, so the first fit fails
- <b style="color:#73000A;">B.</b> At $\lambda = 1000$ the useless slopes are exactly zero and the useful
  ones are untouched
- <b style="color:#73000A;">C.</b> The three fits are identical, since $\lambda$ only rescales the penalty
- <b style="color:#73000A;">D.</b> At $\lambda = 0$ ridge is least squares, and as $\lambda$ grows the slopes
  shrink toward zero without ever arriving. Cross validation is how you
  pick between them
