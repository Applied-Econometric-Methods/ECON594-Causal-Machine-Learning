### Question 1: Clinics That Raise Mortality?

The clinic villages average $16.5$ infant deaths per thousand and the rest average
$12.4$, a naive difference of $+4.1$. Matching each clinic village to the untreated
village with the closest score gives an ATT of $-7$.

What explains the flip in sign?

- <b style="color:#73000A;">A.</b> Matching adds noise, and with nine villages the sign can go either way
- <b style="color:#73000A;">B.</b> The clinics really did raise mortality, and matching hides it by dropping villages 7, 8 and 9
- <b style="color:#73000A;">C.</b> The NGO put clinics in the poorest villages, and poorer villages have higher infant mortality with or without a clinic
- <b style="color:#73000A;">D.</b> The naive difference is an ATE and matching gives an ATT, and those always have opposite signs

### Question 2: Village 7's Gap

Village 7 has no clinic and an infant mortality rate of $4$. Its closest treated
village by score is village 1, with a rate of $10$.

What does village 7 add to the matching estimator?

$$\widehat{ATE} = \frac{1}{N}\sum_{i=1}^{N}\Big\{\big(Y_i - Y_{jm}(i)\big)T_i + \big(Y_{jm}(i) - Y_i\big)(1 - T_i)\Big\}$$

- <b style="color:#73000A;">A.</b> $4 - 10 = -6$, its own outcome minus its match
- <b style="color:#73000A;">B.</b> $0$, because untreated villages only serve as matches and never get a gap of their own
- <b style="color:#73000A;">C.</b> $10$, the outcome it borrows from village 1
- <b style="color:#73000A;">D.</b> $10 - 4 = +6$. It is untreated, so the second term is on: the treated match minus its own outcome

### Question 3: Should We Report the ATE?

Matching both ways gives an ATT of $-7$, an ATU of $-1$ and an ATE of $-3.67$.
Villages 7, 8 and 9 all borrow village 1, about $0.4$ away in score.

Which is the best reading?

- <b style="color:#73000A;">A.</b> The ATE is the better number to report, since it uses all nine villages
- <b style="color:#73000A;">B.</b> The ATT is credible. The ATE leans on three villages whose only match is far away, and their gaps of $+6$, $+2$ and $+4$ are what pull it toward zero
- <b style="color:#73000A;">C.</b> Clinics help treated villages much more than untreated ones, so they should not be built in rich villages
- <b style="color:#73000A;">D.</b> The ATU is the credible number, because untreated villages are the majority

### Question 4: The Benchmark Falls

In the management training data, the naive difference in engagement between
trained and untrained managers is about $0.43$. Regression with the covariates
gives about $0.27$.

What does the drop tell us?

- <b style="color:#73000A;">A.</b> The naive number had positive bias. Managers whose teams were already engaged were more likely to take the training
- <b style="color:#73000A;">B.</b> The training's effect shrank over time, and regression picks up the later, smaller effect
- <b style="color:#73000A;">C.</b> Regression is biased toward zero, so the true effect is probably closer to $0.43$
- <b style="color:#73000A;">D.</b> The covariates were measured after the training, so regression controls away part of the effect

### Question 5: Matching on the Score Instead of X

We could match managers directly on their covariates, or on the propensity score.

Which source of bias does switching to the score remove?

- <b style="color:#73000A;">A.</b> All of it. Equal scores mean equal covariates, so there is no leak left
- <b style="color:#73000A;">B.</b> The bias from thin overlap, where the nearest match is still far away
- <b style="color:#73000A;">C.</b> The bias from a logit that is the wrong model of who gets treated
- <b style="color:#73000A;">D.</b> The bias that grows with the number of covariates, since we now match on a single number

### Question 6: Two Managers at 0.2

Two managers both have $\hat{e}(X) = 0.2$. One took the training and one did not.

What are their inverse propensity weights?

- <b style="color:#73000A;">A.</b> $5$ for both, since they have the same score
- <b style="color:#73000A;">B.</b> $0.2$ for the trained manager and $0.8$ for the other
- <b style="color:#73000A;">C.</b> $1/0.2 = 5$ for the trained manager and $1/0.8 = 1.25$ for the other. The trained one looks like the untrained, so it gets the big weight
- <b style="color:#73000A;">D.</b> $1.25$ for the trained manager and $5$ for the other, since untrained managers are the more common kind at this score

### Question 7: A Very Good Predictor of Training

What happens if we add it to the propensity model?

- <b style="color:#73000A;">A.</b> Bias falls, because the score now predicts treatment better
