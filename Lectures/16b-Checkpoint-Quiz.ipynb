### Question 1: From CATE to ATE

A restaurant chain finds that 70% of its days have a CATE of $12$ extra sales per unit
of discount, and the other 30% have a CATE of $2$.

What is the ATE?

- <b style="color:#73000A;">A.</b> $7$, the average of $12$ and $2$
- <b style="color:#73000A;">B.</b> $9$
- <b style="color:#73000A;">C.</b> $12$, the CATE of the larger group
- <b style="color:#73000A;">D.</b> It cannot be computed without each day's individual effect

### Question 2: Two Identical Customers

Two customers have exactly the same characteristics $x$, and the model's CATE for them
is $\tau(x) = 8$.

Which statement is right?

- <b style="color:#73000A;">A.</b> Each of them will spend exactly $8$ more if treated
- <b style="color:#73000A;">B.</b> Their individual effects must average to exactly $8$
- <b style="color:#73000A;">C.</b> The model must be wrong, because two different people cannot share an effect
- <b style="color:#73000A;">D.</b> Both get the same estimated effect of $8$, but their individual effects can differ, and we never see either one

### Question 6: Everyone Ends Up Here

On the cumulative effect curve, the random model, the ML model and the regression CATE
all end at the same height, about $32$, at $100\%$.

Why?

- <b style="color:#73000A;">A.</b> At 100% every model's group is all the test days, so each one estimates the same slope on the same data: the ATE
- <b style="color:#73000A;">B.</b> All three models are calibrated to the average effect
- <b style="color:#73000A;">C.</b> The curve is normalized so that it ends at the ATE
- <b style="color:#73000A;">D.</b> Errors in the three models cancel out on average

### Question 8: One Number per Model

Which reading is right?

- <b style="color:#73000A;">A.</b> ML orders effects best, since it is the most accurate predictor of sales

- <b style="color:#73000A;">C.</b> Regression CATE orders effects best, and on this data ML does worse than random
- <b style="color:#73000A;">D.</b> The AUC says how close each model's predicted CATEs are to the true CATEs
