### Question 1: The Model That Never Saw the Price

The denoise model $M_y$ predicts sales from temp, weekday and cost. Price is not
one of its inputs. It is fit on the same rows it then predicts, no cross
prediction, and it is flexible enough to fit every day's sales almost perfectly.

What happens to the final regression of the sales residual on the price residual?

- <b style="color:#73000A;">A.</b> Nothing. $M_y$ never saw the price, so it cannot touch the price effect
- <b style="color:#73000A;">B.</b> The slope shrinks towards zero. $M_y$ memorised each day's sales, including the part caused by that day's price surprise, so that part is gone from the residual
- <b style="color:#73000A;">C.</b> The slope keeps its size but the standard error explodes
- <b style="color:#73000A;">D.</b> The slope flips sign, back to the raw +1.23

### Question 2: Negative Effect, Which Way Does the Curve Go?

On the random test set the price elasticity is $-2.55$: higher prices, lower
sales. You rank the test days by a model's predicted effect, largest first, and
draw the normalized cumulative gain curve.

For a good model, what does the curve look like?

- <b style="color:#73000A;">A.</b> Below zero, with a negative AUC, because the effect is negative
- <b style="color:#73000A;">B.</b> Above zero, with a positive AUC. The days ranked first have effects above the average, closer to zero than $-2.55$
- <b style="color:#73000A;">C.</b> Flat at zero. The baseline already absorbs the ATE
- <b style="color:#73000A;">D.</b> It depends on the sign of the ATE, so you cannot tell without the raw curve

### Question 3: Trying a Price

The non-parametric final model was fit to predict the sales residual from
temp, weekday, cost and the <b style="color:#73000A;">price residual</b>. For a day
whose debias model prediction is $\hat{M}_t(X_i) = 6$, you want the predicted
sales residual at a price of 8.

What goes into the model's price slot?

- <b style="color:#73000A;">A.</b> 8, the price you want to try
- <b style="color:#73000A;">B.</b> 2, the trial price minus what a day like this usually charges
- <b style="color:#73000A;">C.</b> 6, the usual price, since the model is conditional on it
- <b style="color:#73000A;">D.</b> 8 minus the day's actual price, whatever was charged that day
